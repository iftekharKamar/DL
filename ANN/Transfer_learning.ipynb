```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

import cv2
import os
import PIL.Image as Image

import tensorflow_hub as hub
```

```python
classifier = tf.keras.Sequential([
    hub.KerasLayer(
        "https://tfhub.dev/google/tf2-preview/mobilenet_v2/classification/4",
        input_shape=(224, 224, 3),
        trainable=False
    )
])

classifier.build([None, 224, 224, 3])
classifier.summary()
```

```python
def preprocess_image(image_path):
    img = Image.open(image_path).convert('RGB')
    img = img.resize((224, 224))
    img_array = np.array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)
    return img_array
```

```python
val=preprocess_image(r"C:\Users\Iftekhar\OneDrive\Desktop\Iftekhar\DL\ANN\sunflower.png")
```

```python
res=classifier.predict(val)
```

```python
pricted_class = np.argmax(res, axis=-1)
print("Predicted class:", pricted_class)
```
